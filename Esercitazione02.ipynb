```python
#Esercizio:
#Creare una classe Person , estenderla con due sottoclassi Dipendente e Studente
#Estendere Dipendenti con 2 sottoclassi Docente e Freelance

#Una classe deve essere astratta e ogni classe deve avere delle proprietà specializzanti

from abc import ABC, abstractmethod

class Person(ABC):
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    @abstractmethod
    def info(self):        
        pass
    
class Studente(Person):
    def __init__(self, name, age, matricola):
        super().__init__(name,  age)
        self.matricola = matricola
        
    def info(self):
        return f"Studente: {self.name}, Età: {self.age}, Matricola: {self.matricola}"
    
class Dipendente(Person, ABC):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary
        
class Docente(Dipendente):
    def __init__(self, name, age, salary, subject):
        super().__init__(name, age, salary)
        self.subject = subject
        
    def info(self):
        return f"Docente: {self.name}, Età: {self.age}, Salario: {self.salary}, Materia: {self.subject}"
    
class Freelance(Dipendente):
    def __init__(self, name, age, salary, partita_iva):
        super().__init__(name, age, salary)
        self.partita_iva = partita_iva
        
    def info(self):
        return f"Freelance: {self.name}, Età: {self.age}, Salario: {self.salary}, Partita IVA: {self.partita_iva}"
```
